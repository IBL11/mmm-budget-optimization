## Feature Engineering Notebook

### Overview
This notebook performs feature engineering on the cleaned marketing dataset. Feature engineering is a critical step in Marketing Mix Modelling as it transforms raw business variables into meaningful analytical features that better capture consumer behaviour, advertising effectiveness, market conditions, and temporal effects.

The engineered features developed in this notebook are designed to support the measurement of channel performance, identify diminishing returns, capture delayed marketing effects, and improve the predictive performance of the marketing mix model.

### Objectives
The key objectives of this notebook are to:

- Load the cleaned dataset from the `data/processed` directory.
- Create time-based features from the month variable.
- Generate marketing channel contribution metrics.
- Calculate customer engagement measures.
- Create advertising efficiency indicators.
- Develop ROI and Return on Ad Spend (ROAS) metrics.
- Create lagged advertising features to capture delayed marketing effects.
- Generate rolling average spending indicators.
- Construct composite business and competition metrics.
- Prepare the dataset for Marketing Mix Modelling and budget optimization.
- Save the feature-engineered dataset to the `data/processed` directory.

### Engineered Feature Categories

#### 1. Time-Based Features
Features derived from the date variable to capture seasonal and temporal patterns.

Examples:
- Year
- Quarter
- Month Number

#### 2. Marketing Effectiveness Features
Measures that quantify the effectiveness of marketing campaigns.

Examples:
- Click-Through Rate (CTR)
- Conversion Rate
- Revenue per Conversion

#### 3. Advertising Investment Features
Metrics that represent expenditure allocation across channels.

Examples:
- TV Spend Share
- Digital Spend Share
- Radio Spend Share
- Outdoor Spend Share

#### 4. ROI Features
Financial indicators used to evaluate marketing performance.

Examples:
- Return on Ad Spend (ROAS)
- Cost per Conversion
- Revenue per Unit Sold

#### 5. Delayed Impact Features
Variables designed to capture the carry-over effects of advertising activities.

Examples:
- Lag 1 Total Advertising Spend
- Lag 2 Total Advertising Spend
- Three-Month Rolling Average Spend

#### 6. Market Influence Features
Features representing competitive and promotional market conditions.

Examples:
- Competitive Advantage Score
- Promotion Impact Score
- Marketing Pressure Index

### Expected Output
The feature-engineered dataset generated from this notebook will be saved as:

`data/processed/feature_engineered_marketing_mix_data.csv`

This dataset will be used as the primary input for the Marketing Mix Modelling, channel effectiveness analysis, ROI evaluation, and budget optimization stages of the project.


```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    "../data/processed/cleaned_marketing_mix_data.csv"
)
```

```python
df['month'] = pd.to_datetime(df['month'])
```

Date Features

```python
df['year'] = df['month'].dt.year

df['quarter'] = df['month'].dt.quarter

df['month_number'] = df['month'].dt.month
```

Spend Share Features

```python
df['tv_share'] = (
    df['tv_spend_lkr']
    / df['total_ad_spend_lkr']
)

df['digital_share'] = (
    (
        df['facebook_instagram_spend_lkr']
        + df['google_search_spend_lkr']
        + df['youtube_spend_lkr']
    )
    / df['total_ad_spend_lkr']
)

df['radio_share'] = (
    df['radio_spend_lkr']
    / df['total_ad_spend_lkr']
)

df['outdoor_share'] = (
    df['outdoor_spend_lkr']
    / df['total_ad_spend_lkr']
)
```

Channel CTR Features

```python
df['ctr'] = (
    df['clicks']
    / df['impressions']
)
```

Conversion Rate

```python
df['conversion_rate'] = (
    df['conversions']
    / df['clicks']
)
```

Revenue Per Unit

```python
df['revenue_per_unit'] = (
    df['sales_revenue_lkr']
    / df['units_sold']
)
```

Revenue Per Conversion

```python
df['revenue_per_conversion'] = (
    df['sales_revenue_lkr']
    / df['conversions']
)
```

Return on Ad Spend

```python
df['roas'] = (
    df['sales_revenue_lkr']
    / df['total_ad_spend_lkr']
)
```

Cost Per Conversion

```python
df['cost_per_conversion'] = (
    df['total_ad_spend_lkr']
    / df['conversions']
)
```

Lag Features (Time Delay Effect)

```python
df = df.sort_values('month')

df['lag_1_total_spend'] = (
    df['total_ad_spend_lkr'].shift(1)
)

df['lag_2_total_spend'] = (
    df['total_ad_spend_lkr'].shift(2)
)
```

Rolling Average Spend

```python
df['rolling_3_month_spend'] = (
    df['total_ad_spend_lkr']
    .rolling(3)
    .mean()
)
```

```python
7. ROAS (Return On Ad Spend)

Critical MMM metric.

df['roas'] = (
    df['sales_revenue_lkr']
    / df['total_ad_spend_lkr']
)

8. Cost Per Conversion
df['cost_per_conversion'] = (
    df['total_ad_spend_lkr']
    / df['conversions']
)

9. Lag Features (Time Delay Effect)

Very important since MMM measures carryover impact.

df = df.sort_values('month')

df['lag_1_total_spend'] = (
    df['total_ad_spend_lkr'].shift(1)
)

df['lag_2_total_spend'] = (
    df['total_ad_spend_lkr'].shift(2)
)

10. Rolling Average Spend

Captures advertising carry-over.

df['rolling_3_month_spend'] = (
    df['total_ad_spend_lkr']
    .rolling(3)
    .mean()
)

11. Marketing Pressure Index

Combining all major paid channels.

df['marketing_pressure'] = (
    df['tv_spend_lkr']
    + df['facebook_instagram_spend_lkr']
    + df['google_search_spend_lkr']
    + df['youtube_spend_lkr']
)

12. Competitive Advantage Score

MMM usually includes competitor actions.

df['competitive_advantage'] = (
    100 - df['competitor_pressure_index_0_100']
)

13. Promotion Impact Score
df['promotion_impact'] = (
    df['promotion_discount_pct']
    * df['seasonal_demand_index']
)

14. Fill New Lag Nulls
df.fillna(0, inplace=True)
```
